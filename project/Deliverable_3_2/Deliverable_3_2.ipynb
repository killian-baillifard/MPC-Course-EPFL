```python
%load_ext autoreload
%autoreload 2

# Get parent directory and add to sys.path
import os
import sys

parent_dir = os.path.dirname(os.getcwd())
sys.path.append(parent_dir)

# Require ipympl
%matplotlib widget 
```

```python
# MPC import
import numpy as np
from LinearMPC.MPCVelControl import MPCVelControl
from src.rocket import Rocket
from src.vel_rocket_vis import RocketVis

rocket_obj_path = os.path.join(parent_dir, "Cartoon_rocket.obj")
rocket_params_path = os.path.join(parent_dir, "rocket.yaml")
```

```python
Ts = 0.05
sim_time = 7
H = 3.0

rocket = Rocket(Ts=Ts, model_params_filepath=rocket_params_path)
mpc = MPCVelControl().new_controller(rocket, Ts, H)
```

```python
x0 = np.zeros(12)
x_target = np.array([
    0.0, 0.0, 0.0,	# omega
    0.0, 0.0, np.deg2rad(35),	# r
    3.0, 3.0, 3.0,	# v
    0.0, 0.0, 0.0	# p
])

t_cl, x_cl, u_cl, t_ol, x_ol, u_ol, _ = rocket.simulate_control(
    mpc, sim_time, H, x0, x_target=x_target, method="linear"
)
```

```python
vis = RocketVis(rocket, rocket_obj_path)
vis.anim_rate = 1.0
vis.animate(
    t_cl[:-1],
    x_cl[:, :-1],
    u_cl,
    Ref=x_target.reshape(-1, 1).repeat(u_cl.shape[1], axis=1),
    T_ol=t_ol[..., :-1],
    X_ol=x_ol,
    U_ol=u_ol,
)
```

```python
from plotter import plot_traj

x_target = np.tile(x_target.reshape(12, 1), (1, x_cl.shape[1]))
plot_traj('Origin to 3 ms in x, y, z and 35° roll', x_cl, u_cl, t_cl, x_target)
```
